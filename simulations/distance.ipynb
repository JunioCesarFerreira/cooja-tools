```python
import math

def dist(p1, p2):
    """Calcula a distância euclidiana entre dois pontos p1 e p2 no plano."""
    if len(p1) != 2 or len(p2) != 2:
        raise ValueError("Cada ponto deve ter exatamente duas coordenadas (x, y).")
    return math.dist(p1, p2)  # disponível a partir do Python 3.8
```

```python
ponto1 = [
          450.6578947368421,
          198.02631578947367
        ]
ponto2 = [
          467.7631578947368,
          180.26315789473685
        ]

d = dist(ponto1, ponto2)
print(f"dist= {d:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def path_loss(d, Pt=-30, n=2.5, d0=1.0, sigma=2.0):
    """
    Modelo log-distance com ruído normal.
    Pt: potência transmitida (dBm)
    n: expoente de perda
    d0: distância de referência (m)
    sigma: desvio padrão do ruído (dB)
    """
    np.random.seed(42)
    X_sigma = np.random.normal(0, sigma, size=len(d))
    Pr = Pt - 10 * n * np.log10(d / d0) + X_sigma
    return Pr

# distâncias de 1 a 100 m
dist = np.linspace(1, 100, 100)
rssi = path_loss(dist, Pt=-15, n=4.0, sigma=0)

plt.figure(figsize=(7, 5))
plt.plot(dist, rssi, label="RSSI simulado", color="blue")
plt.xlabel("Distância (m)")
plt.ylabel("RSSI (dBm)")
plt.title("Modelo Log-Distance Path Loss com Ruído")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
